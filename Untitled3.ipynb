# Variáveis e Tipos

```python
nome = input("Qual seu nome? ")
idade = int(input("Digite sua idade: "))  # converte pra número
estudando = input("Você está estudando? (sim/não): ").lower() == "sim"

print(f"Nome: {nome}")
print(f"idade: {idade}")
print(f"Está estudando? {estudando}")
```

```python
num1 = float(input("Digite um número: "))
num2 = float(input("Digite outro número: "))
soma = num1 + num2
diferença = num1 - num2
media = (num1 + num2) / 2
multiplicação = num1 * num2

print(f"soma: {soma}")
print(f"diferença: {diferença}")
print(f"media: {media}")
print(f"multiplicação: {multiplicação}")
```

```python
valor_em_reais = float(input("Digite um valor em reais: "))
cotacao_dolar = float(input("Digite o valor da cotação do dólar: "))
valor_em_dolares = valor_em_reais / cotacao_dolar

print(f"Você teria {valor_em_dolares:.2f} dólares.")
```

```python
num1 = float(input("Digite um número: "))
dobro = num1 * 2
triplo = num1 * 3
raiz_quadrada = num1 ** 0.5

print(f"Dobro: {dobro}")
print(f"Triplo: {triplo}")
print(f"Raiz quadrada: {raiz_quadrada}")
```

# Condições com IF, ELSE e ELIF


```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = (nota1 + nota2) / 2

if media >= 8:
  print("✅ Aprovado")
elif media >= 6:
  print("🟡 Recuperação")
else:
  print("❌ Reprovado")
```

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

if num1 > num2:
    print(f"O número {num1} é maior.")
elif num2 > num1:
    print(f"O número {num2} é maior.")
else:
    print("Os números são iguais.")
```

```python
ano = int(input("Digite um ano: "))

if ano % 4 == 0:
    print(f"O ano {ano} é bissexto!")
else:
    print(f"O ano {ano} NÃO é bissexto!")

```

```python
idade = int(input("Digite sua idade: "))

if idade < 18:
    print("Menor de idade")
elif idade < 60:
    print("Maior de idade")
else:
    print("Idoso")
```

```python
peso = float(input("Digite seu peso: "))
altura = float(input("Digite sua altura: "))
imc = peso / (altura ** 2)

print(f"Seu IMC é: {imc:.2f}")

if imc < 18.5:
    print("Abaixo do peso")
elif imc < 25:
    print("Peso normal")
elif imc < 30:
    print("Sobrepeso")
else:
    print("Obesidade")
```

```python
usuario_correto = "admin"
senha_correta = "1234"

usuario = input("Digite o nome de usuário: ")
senha = input("Digite a senha: ")

if usuario == usuario_correto and senha == senha_correta:
    print("✅ Acesso concedido!")
else:
    print("❌ Usuário ou senha inválidos.")

```

```python
usuario_correto = "admin"
senha_correta = "1234"
tentativas = 0

while tentativas < 3:
    usuario = input("Usuário: ")
    senha = input("Senha: ")

    if usuario == usuario_correto and senha == senha_correta:
        print("✅ Acesso concedido!")
        break  # Sai do laço
    else:
        print("❌ Usuário ou senha incorretos.")
        tentativas += 1

if tentativas == 3:
    print("🚫 Número máximo de tentativas atingido. Tente novamente mais tarde.")
```
